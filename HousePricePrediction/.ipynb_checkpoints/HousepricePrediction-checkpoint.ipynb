```python
import pandas as pd 
from sklearn.linear_model import LinearRegression 
from sklearn.model_selection import train_test_split 
from sklearn.metrics import r2_score 
import seaborn as sns 
import numpy as np
```

```python
data=pd.read_csv('HousePricePrediction.csv')
df=pd.DataFrame(data)
df.tail()

df=df.fillna(0)
df.info()
```

```python
sns.scatterplot(y=df['SalePrice'],x=df['YearBuilt'],hue=df['LotArea'])
```

```python
df['BldgType'].value_counts()
```

```python
df['MSZoning']=df['MSZoning'].map({'RL':0,'RM':1,'FV':2,'RH':3})
df['LotConfig']=df['LotConfig'].map({'Inside':0,'Corner':1,'CulDSac':2,'FR2':3,'FR3':4})
df['BldgType']=df['BldgType'].map({
    '1Fam': 0,
    '2fmCon': 1,
    'Duplex': 2,
    'TwnhsE':3,
    'Twnhs':4
})
df['Exterior1st']=df['Exterior1st'].replace({
    'VinylSd': 0,
    'MetalSd': 1,
    'HdBoard': 2,
    'Wd Sdng': 3,
    'Plywood': 4,
    'CemntBd': 5,
    'BrkFace':  6,
    'WdShing':   7,
    'AsbShng':   8,
    'Stucco':  9,
    'BrkComm':   10,
    'AsphShn':   11,
    'Stone':12,
    'CBlock' :   13,
    'ImStucc':  14
})

df[['MSZoning','LotConfig','BldgType','Exterior1st']].isna().sum()
df[df['MSZoning'].isna()]
```

```python
df['MSZoning'] = df['MSZoning'].fillna(0)
df.head()
```

```python
Housepredictor=LinearRegression()
x_train,x_test,y_train,y_test=train_test_split(df[['MSSubClass','MSZoning','LotArea','LotConfig','BldgType','OverallCond','YearBuilt','YearRemodAdd','Exterior1st','BsmtFinSF2','TotalBsmtSF']],df[['SalePrice']],test_size=0.2,random_state=42)
Housepredictor.fit(x_train,y_train)  
priceprediction=Housepredictor.predict(x_test)
accuracy=r2_score(y_test,priceprediction)
accuracy
```
