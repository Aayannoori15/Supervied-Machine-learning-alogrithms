```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler 
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, r2_score, precision_score
```

```python
data=pd.read_csv('Iris.csv')
data.head()
```

```python
data.isna().sum()
```

```python
y=data['Species'].map({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
```

```python
x=data.drop('Species',axis=1)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
model=GaussianNB()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
```

```python
print(f'accuracy score is {accuracy_score(y_test,y_pred)}')
print(f'r2 score is {r2_score(y_test,y_pred)}')
#print(f'precision score is {precision_score(y_test,y_pred)}')
```

```python
knn=KNeighborsClassifier(n_neighbors=5)
knn.fit(x_train,y_train)
knn_pred=knn.predict(x_test)
print(f'accuracy score is {accuracy_score(y_test,knn_pred)}')
print(f'r2 score is {r2_score(y_test,knn_pred)}')
```

```python
lr=LogisticRegression()
lr.fit(x_train,y_train)
lr_pred=lr.predict(x_test)
print(f'accuracy score is {accuracy_score(y_test,lr_pred)}')
print(f'r2 score is {r2_score(y_test,lr_pred)}')
```

```python
step=[('Scaler',StandardScaler()),('knn',KNeighborsClassifier())]
pipeline=Pipeline(step)
param_grid={'knn__n_neighbors':[3,5,7,9,11]}

cv=GridSearchCV(
    pipeline,
    param_grid=param_grid,
    cv=5
)
cv.fit(x_train,y_train)
cv_pred=cv.predict(x_test)
print("Best k:", cv.best_params_)
print("Best CV score:", cv.best_score_)
```
